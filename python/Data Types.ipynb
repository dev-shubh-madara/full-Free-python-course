Step 1: No space left → allocate a BIGGER buffer (new memory block)
Step 2: Copy all existing elements to the new buffer
Step 3: Add the new element
Step 4: Release the old buffer

Old memory:  [ 1 | 2 | 3 ]           ← freed
New memory:  [ 1 | 2 | 3 | 4 | _ | _ | _ | _ ]  ← new id (internally)

```python
# Sets in Python

## Definition
'''

Set ek collection hai jisme sirf **unique** values hoti hain — duplicates automatically remove ho jaate hain.
'''
## Set Banana
a = {1, 2, 3}        # directly
b = set()            # khali set — {} mat use karo, wo dictionary ban jaata hai

## 3 Main Properties

### 1. Unique — No Duplicates
a = {1, 2, 2, 3, 3}
print(a)  # {1, 2, 3}

### 2. Unordered — Koi Fixed Order Nahi
a = {3, 1, 2}
print(a)  # kuch bhi aa sakta hai — order guaranteed nahi
a[0]      # Error! — index nahi hota set mein

### 3. Fast Search — O(1)
3 in {1, 2, 3, 4, 5}   # direct jump karta hai — ek ek check nahi karta

## Internal Working — Hash Table

### Hash Function
hash(value) → ek number deta hai
slot = hash(value) % total_slots    # ye decide karta hai kahan store hoga

hash(1) % 8 = 1  → slot 1
hash(2) % 8 = 2  → slot 2
hash(3) % 8 = 3  → slot 3

### Memory mein
a = {1, 2, 3}
Slot:   [ 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 ]
Value:  [ _ | 1 | 2 | 3 | _ | _ | _ | _ ]
Isliye order nahi hota — hash decide karta hai jagah

### Duplicate Ignore Kaise Hota Hai
a.add(2)  # agar 2 pehle se hai
→ hash(2) % 8 = 2 → slot 2 dekho → already 2 hai → SKIP

### Collision — Jab Do Elements Ka Same Slot Ho
hash(0) % 8 = 0
hash(8) % 8 = 0   # same slot! → Collision
Solution — Probing: agla khali slot dhundho
[ 0 | 8 | _ | _ | _ | _ | _ | _ ]

## Search Speed Comparison
# List → ek ek karke dekho → O(n) → slow
# Set  → seedha sahi slot pe jaao → O(1) → fast

## Common Operations
a = {1, 2, 3}
b = {3, 4, 5}

a | b   # Union        → {1, 2, 3, 4, 5} — dono mila do
a & b   # Intersection → {3}             — sirf common
a - b   # Difference   → {1, 2}          — a mein hai b mein nahi

a.add(4)      # element add karo
a.remove(2)   # element remove karo — error deta hai agar nahi hai
a.discard(2)  # element remove karo — error nahi deta
len(a)        # kitne elements hain

## List vs Set Comparison
#            List              Set
# Duplicate  Allowed           Not Allowed
# Order      Fixed             No Order
# Index      a[0] works        Error
# Search     O(n) slow         O(1) fast
# Internal   Array             Hash Table
# Memory     Kam               Zyada (empty slots hote hain)

## Kab Use Karein
# Set use karo jab:
# → Unique values chahiye ho
# → Fast search chahiye ho (in operator)
# → Duplicates hatane ho list mein se — list(set(a))

## Quick Tip — List se Duplicates Hatao
a = [1, 2, 2, 3, 3, 3]
a = list(set(a))
print(a)  # [1, 2, 3]
```

```python
# ============================================================
# Python List Memory - Summary
# ============================================================

lst = [1, 2, 3]

# ----------------------------------------------------------
# 1. ID check — before and after appending
# ----------------------------------------------------------
print("=== ID Check ===")
print("Before append → id:", id(lst))

for i in range(10000):
    lst.append(i)

print("After 10000 appends → id:", id(lst))
print("Same id? →", id(lst) == id(lst))  # Always True
print("Reason: id() tracks the LIST OBJECT, not internal C buffer\n")

# ----------------------------------------------------------
# 2. Amortized O(1) — capacity growth pattern
# ----------------------------------------------------------
import sys

print("=== Capacity Growth (overallocation) ===")
lst2 = []
prev_size = sys.getsizeof(lst2)

for i in range(20):
    lst2.append(i)
    curr_size = sys.getsizeof(lst2)
    if curr_size != prev_size:
        print(f"len={len(lst2):3d} → size jumped: {prev_size} → {curr_size} bytes (reallocation!)")
        prev_size = curr_size

# ----------------------------------------------------------
# 3. Set vs List — extra slots ka alag reason
# ----------------------------------------------------------
print("\n=== Set vs List ===")
my_list = [1, 2, 3]
my_set  = {1, 2, 3}

print(f"List size: {sys.getsizeof(my_list)} bytes → extra slots for future appends")
print(f"Set  size: {sys.getsizeof(my_set)}  bytes → extra slots to avoid hash collisions")
```
