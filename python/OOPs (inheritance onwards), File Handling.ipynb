```python
# 1. Single Inheritance
# https://media.geeksforgeeks.org/wp-content/uploads/Single.jpg
# Concept:
# One child class inherits from a single parent class.

# Simple Example of Single Inheritance
class A:
    def abc(self):
        print("Hello from class A")

class B(A):
    def xyz(self):
        print("Hello from class B")

b = B()
b.xyz()  # b has access to its own method
b.abc()  # b also has access to method from class A via inheritance

a = A()
a.abc()  # a can only access its own method
# a.xyz()  # Error: 'A' object has no attribute 'xyz'
```

```python
# Real time example of Single Inheritance
# Problem Statement:
# You want to create a simple Employee Management System where you have a base Employee class and a Manager class that inherits from Employee.

class Employee:
    def __init__(self, name, salary): # (name="Avika", salary=80000)
        self.name = name
        self.salary = salary

    def show_details(self):
        print(f"Employee Name: {self.name}, Salary: ₹{self.salary}")
        
class Manager(Employee):
    def __init__(self, name, salary, department): # (name="Avika", salary=80000, department="HR")
        super().__init__(name, salary)  # Call parent constructor. (name="Avika", salary=80000)
        self.department = department

    def show_details(self):
        super().show_details()  # Call parent method
        print(f"Department: {self.department}")

# Using the classes
emp = Employee("Deepak", 60000)
emp.show_details()

mgr = Manager("Avika", 80000, "HR")
mgr.show_details()  # Manager can access Employee methods and attributes

# What is super()?
# super() is a built-in function in Python that allows you to call methods from a parent class.
# It is commonly used in inheritance to access parent class methods and attributes from a child class.
# Using super() helps a child class to avoid directly referencing the parent class name, making the code more maintainable and flexible, especially in complex inheritance hierarchies.
```

```python
# 2. Multi-level Inheritance
# https://media.geeksforgeeks.org/wp-content/uploads/multi.jpg
# Concept:
# A child class inherits from a parent class, which itself inherits from another parent class.
# eg: child -> parent -> grandparent

# Simple Example of Multi-level Inheritance
class Grandfather:
    def show_grandfather(self):
        print("Hello from Grandfather")

class Father(Grandfather):
    def show_father(self):
        print("Hello from Father")

class Son(Father):
    def show_son(self):
        print("Hello from Son")
        
s = Son()
s.show_son()            # Hello from Son
s.show_father()        # Hello from Father
s.show_grandfather()   # Hello from Grandfather
```

```python
# Real World Example of Multi-level Inheritance
# Problem Statement:
# You want to create a Vehicle class hierarchy where you have a base Vehicle class, a Car class that inherits from Vehicle, and a SportsCar class that inherits from Car.

class Vehicle:
    def __init__(self, brand, model): # (brand="Toyota", model="Corolla")
        self.brand = brand # self.brand = Toyota
        self.model = model # self.model = Corolla

    def show_details(self):
        print(f"Vehicle Brand: {self.brand}, Model: {self.model}")

class Car(Vehicle):
    def __init__(self, brand, model, doors): # (brand="Honda", model="Civic", doors=4)
        super().__init__(brand, model)  # Call parent constructor
        self.doors = doors

    def show_details(self):
        super().show_details()  # Call parent method
        print(f"Number of Doors: {self.doors}")

class SportsCar(Car):
    def __init__(self, brand, model, doors, top_speed): # (brand="Ferrari", model="488", doors=2, top_speed=211)
        super().__init__(brand, model, doors)  # Call parent constructor
        self.top_speed = top_speed

    def show_details(self):
        super().show_details()  # Call parent method
        print(f"Top Speed: {self.top_speed} mph")

# Using the classes
vehicle = Vehicle("Toyota", "Corolla")
vehicle.show_details()

car = Car("Honda", "Civic", 4)
car.show_details()

sports_car = SportsCar("Ferrari", "488", 2, 211)
sports_car.show_details()
```

```python
# 3. Hierarchical Inheritance
# https://media.geeksforgeeks.org/wp-content/uploads/20200108144949/Hierarchical-inheritance1.png
# Concept:
# Multiple child classes inherit from a single parent class.

# Simple Example of Hierarchical Inheritance
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meows")

class Cow(Animal):
    def moo(self):
        print("Cow moos")

dog = Dog()
dog.bark()   # Dog barks
dog.speak()  # Animal speaks

cat = Cat()
cat.meow()   # Cat meows
cat.speak()  # Animal speaks

cow = Cow()
cow.moo()    # Cow moos
cow.speak()  # Animal speaks
# cow.bark()   # Error: 'Cow' object has no attribute 'bark'
```

```python
# Real World Example of Hierarchical Inheritance
# Problem Statement:
# You want to create a simple content management system where you have a base Content class and multiple child classes like Article, Video, and Podcast that inherit from Content.

class Content:
    def __init__(self, title, author): 
        self.title = title
        self.author = author

    def display_info(self):
        print(f"Title: {self.title}, Author: {self.author}")

class Article(Content):
    def __init__(self, title, author, word_count):
        super().__init__(title, author)
        self.word_count = word_count

    def display_info(self):
        super().display_info()
        print(f"Word Count: {self.word_count}")

class Video(Content):
    def __init__(self, title, author, duration):
        super().__init__(title, author)
        self.duration = duration

    def display_info(self):
        super().display_info()
        print(f"Duration: {self.duration}")

class Podcast(Content):
    def __init__(self, title, author, length):
        super().__init__(title, author)
        self.length = length

    def display_info(self):
        super().display_info()
        print(f"Length: {self.length}")


# Using the classes
article = Article("OOP in Python", "Alice", 1500)
article.display_info()

video = Video("Python Tutorial", "Bob", 10)
video.display_info()

podcast = Podcast("Tech Talks", "Charlie", 30)
podcast.display_info()
```

```python
# 4. Multiple Inheritance
# https://binaryterms.com/wp-content/uploads/2020/06/Multiple-Inheritance-in-Object-Orientation.jpg
# Concept:
# A child class inherits from more than one parent class.

# Simple Example of Multiple Inheritance
class Father:
    def show_father(self):
        print("Hello from Father")

class Mother:
    def show_mother(self):
        print("Hello from Mother")

class Child(Father, Mother):
    def show_child(self):
        print("Hello from Child")

c = Child()
c.show_child()       # Hello from Child
c.show_father()     # Hello from Father
c.show_mother()     # Hello from Mother
```

```python
# Real World Example of Multiple Inheritance 
# Problem Statement:
# IT Employee with Multiple Roles

# Parent 1: Developer
class Developer:
    def code(self):
        print("Develops backend and frontend features.")

# Parent 2: Tester
class Tester:
    def test(self):
        print("Performs functional and integration testing.")

# Child: TechLead (inherits from both)
class TechLead(Developer, Tester):
    def manage(self):
        print("Manages the software development lifecycle.")

lead = TechLead()

lead.code()     # from Developer
lead.test()     # from Tester
lead.manage()   # from TechLead

# The TechLead class inherits skills from both Developer and Tester, demonstrating Multiple Inheritance.
```

```python
# Understanding Method Resolution Order (MRO)
# In Multiple Inheritance, if both parent classes have a method with the same name, Python decides which one to call first based on MRO (Method Resolution Order).

class Developer:
    def work(self):
        print("Writes code.")

class Tester:
    def work(self):
        print("Tests software.")

class TechLead(Developer, Tester):
    pass

lead = TechLead()
lead.work()

# Python picked the method from the first parent (Developer) since Developer was listed first in the inheritance list.
```

```python
# To check the order:
print(TechLead.__mro__)

# This is how Python resolves method conflicts.
```

```python
# 5. Hybrid Inheritance
# https://media.geeksforgeeks.org/wp-content/uploads/20231020172409/Hybrid-Inheritance-in-CPP.png
# Concept:
# A combination of two or more types of inheritance.
# Simple Example of Hybrid Inheritance
class A:
    def method_a(self):
        print("Method from class A")

class B(A):
    def method_b(self):
        print("Method from class B")

class C(A):
    def method_c(self):
        print("Method from class C")

class D(B, C):
    def method_d(self):
        print("Method from class D")

d = D()
d.method_d()  # Method from class D
d.method_b()  # Method from class B
d.method_c()  # Method from class C
d.method_a()  # Method from class A

# Hybrid Inheritance = a combination of two or more inheritance types (like single, multiple, multilevel, or hierarchical).
# It’s called “hybrid” because it blends different inheritance relationships in the same program.
```

| Class      | Description                                           | Type of Inheritance            |
| ---------- | ----------------------------------------------------- | ------------------------------ |
| `Employee` | Base class – defines common attributes                | Base                           |
| `Manager`  | Inherits from Employee (adds managerial capability)   | Single                         |
| `Trainer`  | Independent class – represents training functionality | Independent                    |
| `TechLead` | Inherits from both Manager and Trainer                | Hybrid (Multiple + Multilevel) |


```python
# Training & Management System

# Base Class
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def show_details(self):
        print(f"Employee Name: {self.name}, ID: {self.emp_id}")


# Derived Class 1 (Single Inheritance)
class Manager(Employee):
    def __init__(self, name, emp_id, team_size):
        super().__init__(name, emp_id)
        self.team_size = team_size

    def manage_team(self):
        print(f"{self.name} manages a team of {self.team_size} members.")


# Independent Class
class Trainer:
    def __init__(self, subject):
        self.subject = subject

    def conduct_training(self):
        print(f"Conducts training in {self.subject}.")


# Hybrid Class (Multiple + Multilevel Inheritance)
class TechLead(Manager, Trainer):
    def __init__(self, name, emp_id, team_size, subject):
        Manager.__init__(self, name, emp_id, team_size)
        Trainer.__init__(self, subject)

    def show_role(self):
        print(f"{self.name} is a Tech Lead managing {self.team_size} people and training in {self.subject}.")

lead = TechLead("Neha", "TL101", 8, "Python & AI")

lead.show_details()      # from Employee
lead.manage_team()       # from Manager
lead.conduct_training()  # from Trainer
lead.show_role()         # from TechLead

# This is a hybrid inheritance structure:
# Manager → Employee (Multilevel)
# TechLead → Manager + Trainer (Multiple)
# Combined = Hybrid Inheritance
```

```python
# Module 6 — Polymorphism
# Q: Imagine an e-commerce platform like Amazon.
# When a customer clicks “Pay Now”, the system could process payment via Credit Card, UPI, or PayPal.
# The action name is the same — make_payment() — but the behavior depends on the chosen payment type.
# So, same method name, but different actions.

# Poly - "many"
# Morph - "forms"
# Polymorphism = "many forms"
# Polymorphism allows methods to do different things based on the object calling them, even if they share the same name.
```

### Types of Polymorphism

| Type                      | Description                                                     | Python Support                                                |
| ------------------------- | --------------------------------------------------------------- | ------------------------------------------------------------- |
| **Compile-time (Static)** | Same method name with different signatures (method overloading) |  Not directly supported (Python allows default args instead) |
| **Run-time (Dynamic)**    | Same method name re-defined in child class (method overriding)  |  Fully supported                                             |

```python
# Compile-time (Static) Polymorphism
# It refers to the ability to define multiple methods with the same name but different signatures (different number or types of parameters) within the same class.
# This allows the method to behave differently based on the arguments passed during the method call.
# However, Python does not support method overloading in the traditional sense like some other languages (e.g., Java, C++).
# Instead, Python allows default arguments and variable-length arguments to achieve similar functionality.

# Simple Example of Compile-time Polymorphism
class MathOperations:
    def add(self, a, b, c=0):
        return a + b + c

    def multiply(self, a, b):
        return a * b
    
math = MathOperations()
print(math.add(2, 3))        # Output: 5 (uses 2 parameters)
print(math.add(2, 3, 4))     # Output: 9 (uses 3 parameters)
print(math.multiply(2, 3))   # Output: 6 (uses 2 parameters)
# Here, the add method behaves differently based on the number of arguments passed.
# Hence, it demonstrates compile-time polymorphism.
```

```python
# Runtime (Dynamic) Polymorphism
# It refers to the ability of different classes to provide different implementations of the same method.
# This is typically achieved through method overriding, where a subclass provides a specific implementation of a method that is already defined in its superclass.
# The method to be executed is determined at runtime based on the object type.
# Simple Example of Runtime Polymorphism
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

# Runtime Polymorphism in Action
def make_animal_sound(animal):
    animal.sound()

dog = Dog()
cat = Cat()

make_animal_sound(dog)  # Output: Dog barks
make_animal_sound(cat)  # Output: Cat meows
# Here, the sound method behaves differently based on the object type (Dog or Cat) passed to the make_animal_sound function.
# Hence, it demonstrates runtime polymorphism.
```

```python
# Payment System (Real-World Run-Time Polymorphism)

class Payment:
    def make_payment(self):
        print("Processing generic payment...")

class CreditCard(Payment):
    def make_payment(self):
        print("Processing payment via Credit Card.")

class UPI(Payment):
    def make_payment(self):
        print("Processing payment via UPI.")

class PayPal(Payment):
    def make_payment(self):
        print("Processing payment via PayPal.")

p1 = CreditCard()
p2 = UPI()
p3 = PayPal()

for method in (p1, p2, p3):
    method.make_payment()

# All classes share the same method name make_payment(),
# but each provides its own implementation — this is method overriding polymorphism.
```

### Why Do We Need Abstraction?

| Problem in Procedural / Non-abstracted Design | How Abstraction Helps                                   |
| --------------------------------------------- | ------------------------------------------------------- |
| Too many details → hard to understand         | Hides internal logic, shows simple interface            |
| Duplicate code across classes                 | Common abstract base class can define shared blueprints |
| Changes break dependent code                  | Clients depend on interface not implementation          |
| Security risk (data exposure)                 | Restricts direct access to sensitive functions          |

```python
# Simple example
from abc import ABC, abstractmethod
class Grandfather(ABC):
    
    def buy_bungalow(self):
        print("Bungalow purchased.")

    @abstractmethod
    def buy_penthouse(self):
        pass

    @abstractmethod
    def buy_mercedes(self):
        pass

class Father_S1(Grandfather):
    
    def buy_penthouse(self):
        print("Penthouse purchased.")

    @abstractmethod
    def buy_mercedes(self):
        pass  # Not implemented

    def buy_fortuner(self):
        print("Fortuner purchased.")

class Son_S2(Father_S1):
    
    def buy_mercedes(self):
        print("Mercedes purchased.")

    def buy_penthouse(self):
        print("Penthouse purchased by Son.")

    def buy_bike(self):
        print("Bike purchased.")

class Son(Father_S1):
    
    def buy_mercedes(self):
        print("Mercedes purchased by Son.")

son = Son()
son.buy_bungalow()   # Bungalow purchased.
son.buy_penthouse()  # Penthouse purchased by Son.
son.buy_mercedes()   # Mercedes purchased by Son.   
```

```python
# Abstraction in Python Using abc Module
# Python provides the abc ( Abstract Base Classes ) module for abstraction.

# Example 1 — Payment System
from abc import ABC, abstractmethod

# Abstract Class
class PaymentGateway(ABC):

    def show(self): # Bungalow
        print("Welcome to the Payment Gateway")

    @abstractmethod
    def make_payment(self, amount): # Penthouse
        pass

    @abstractmethod
    def refund(self, amount): # Mercedes
        pass
# The above class defines what must be implemented, but not how.
```

```python
# Now we create subclasses that implement these methods.
class CreditCardPayment(PaymentGateway):
    def make_payment(self, amount): # Penthouse
        print(f"Paid ₹{amount} using Credit Card.")

    def refund(self, amount): # Mercedes
        print(f"Refunded ₹{amount} to Credit Card.")

class UpiPayment(PaymentGateway):
    def make_payment(self, amount): # Penthouse
        print(f"Paid ₹{amount} using UPI.")

    def refund(self, amount): # Mercedes
        print(f"Refunded ₹{amount} to UPI Wallet.")

# Using the classes
cc_payment = CreditCardPayment()
cc_payment.show()
cc_payment.make_payment(5000)
cc_payment.refund(2000)
upi_payment = UpiPayment()
upi_payment.show()
upi_payment.make_payment(3000)
upi_payment.refund(1000)
```

```python
# Module 8 — Special (Magic / Dunder) Methods
# Q: When you write something like print(emp1), how does Python know what to print?
# Or when you type len(my_list), how does it know how to count elements?
# Ans: Because those functions are built into Python
# True — but what actually happens is that Python calls hidden methods like __str__() and __len__() internally!
# These are called Magic Methods (or Dunder Methods, short for “Double Underscore”).
```

```python
# What Are Magic / Dunder Methods?
# Magic methods are special predefined methods in Python that start and end with double underscores (__method__).
# They define how objects of a class interact with built-in operations — such as printing, comparison, addition, etc.

# When you write:
x = [1, 2, 3]
print(len(x))
# Python internally executes:
print(x.__len__())
# Similarly:
print(x)
# internally calls:
print(x.__str__())
# So, magic methods allow us to customize object behavior for native Python operations.
```

### Common Magic Methods and Their Purpose

| Method          | Purpose                                              |
| --------------- | ---------------------------------------------------- |
| `__init__()`    | Constructor – initializes an object                  |
| `__str__()`     | Returns a user-friendly string representation        |
| `__repr__()`    | Returns a developer-readable representation          |
| `__len__()`     | Defines behavior for `len()`                         |
| `__add__()`     | Defines behavior for the `+` operator                |
| `__eq__()`      | Defines equality check (`==`)                        |
| `__lt__()`      | Defines less-than comparison (`<`)                   |
| `__getitem__()` | Defines behavior for accessing elements (`obj[key]`) |


```python
# Example 1 — The Constructor: __init__()
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

emp = Employee("Ravi", 85000)
print(emp.name)
print(emp.salary)
# Here, __init__() is a magic method that initializes the object when created.
```

```python
# Example 2 — String Representation: __str__() and __repr__()
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def __str__(self):
        return f"Employee({self.name}, ₹{self.salary})"

    def __repr__(self):
        return f"Employee(name='{self.name}', salary={self.salary})"
emp = Employee("Ravi", 85000)
print(emp)          # Calls __str__()
print(repr(emp))    # Calls __repr__()
# __str__() is for a user-friendly string representation,
# while __repr__() is for an unambiguous representation useful for debugging.
# Both are magic methods.

# __str__() → meant for user-friendly display.
# __repr__() → meant for developer debugging.
```

```python
# Example 3 — Length: __len__()
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)
    
team = Team(["Alice", "Bob", "Charlie"])
print(team.members)  # Output: ['Alice', 'Bob', 'Charlie']
print(len(team))  # Calls __len__(), Output: 3
# Here, __len__() allows the Team object to respond to len() calls.

# The built-in len() function now works seamlessly on our custom class.
```

```python
# Example 4 — Operator Overloading: __add__()
# Scenario:
# In the real world, you might want to add two Vectors, Matrices, or Financial Reports.
# Let’s see how to make + work for your own class.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(5, 7)
v3 = v1 + v2  # Internally calls v1.__add__(v2)
print(v3)      # Output: Vector(7, 10)
# Here, __add__() defines how to add two Vector objects using the + operator.
```

```python
# Example 5 — Equality Check: __eq__()

class Product:
    def __init__(self, id, price):
        self.id = id
        self.price = price

    def __eq__(self, other):
        return self.id == other.id and self.price == other.price

p1 = Product(101, 250)
p2 = Product(101, 250)
p3 = Product(102, 300)

print(p1 == p2)  # Output: True
print(p1 == p3)  # Output: False
# Here, __eq__() allows us to compare two Product objects using the == operator based on their attributes.
```

```python
# Real-World Example — Bank Account Addition
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def __add__(self, other):
        if isinstance(other, BankAccount): # isinstance - checks if other is a BankAccount instance
            return BankAccount(fusion_account(self.account_number, other.account_number), self.balance + other.balance)
        return NotImplemented

    def __str__(self):
        return f"BankAccount({self.account_number}, {self.balance})"

def fusion_account(acc1, acc2):
    # Simple fusion logic: take the average of the two account numbers
    return (acc1 + acc2) // 2

acc1 = BankAccount(1001, 500)
acc2 = BankAccount(1002, 700)
acc3 = acc1 + acc2  # Internally calls acc1.__add__(acc2)
print(acc3)         # Output: BankAccount(1001, 1200)
# Here, __add__() defines how to add two BankAccount objects using the + operator.
# The fusion_account function combines the account numbers in a simple way.
# This is operator overloading using magic methods in a real-world context.
```

```python
# Composition vs Inheritance
# Q: A Car is a Vehicle, right?
# But it also has an Engine.
# So which concept does that represent — Inheritance or Composition?
# Ans:
# Car is-a Vehicle → Inheritance.
# Car has-a Engine → Composition.
# That’s the essence of this module.
```

| Concept         | Relationship | Example                | Type         |
| --------------- | ------------ | ---------------------- | ------------ |
| **Inheritance** | “is-a”       | `Car` *is-a* `Vehicle` | Hierarchical |
| **Composition** | “has-a”      | `Car` *has-a* `Engine` | Containment  |


```python
# What Is Inheritance (Recap)
# Inheritance means one class derives from another and reuses its attributes and methods.

class Vehicle:
    def start(self):
        print("Vehicle started.")

class Car(Vehicle):
    def drive(self):
        print("Car is driving.")

c = Car()
c.start()   # from Vehicle
c.drive()   # from Car

# Here, Car is-a Vehicle.
# This is hierarchical reuse — good when there’s a logical “is-a” relationship.
```

```python
# What Is Composition
# Composition means one class contains another class’s object and uses it instead of inheriting from it.

class Engine:
    def start(self):
        print("Engine started.")

class Car:
    def __init__(self):
        self.engine = Engine()  # Car has-a Engine

    def drive(self):
        self.engine.start()
        print("Car is driving.")

my_car = Car()
my_car.drive()

# Here, Car has-a Engine.
# Instead of extending Engine, it uses it — that’s composition.
```

| Relationship | Example            | Explanation                           |
| ------------ | ------------------ | ------------------------------------- |
| **is-a**     | “Dog is an Animal” | Inherits general features (move, eat) |
| **has-a**    | “Dog has a Collar” | Uses another class internally         |


### When to Use Composition Instead of Inheritance

| Situation                                                     | Prefer Composition Because…                          |
| ------------------------------------------------------------- | ---------------------------------------------------- |
| The relationship is “has-a,” not “is-a.”                      | Keeps architecture logical and modular.              |
| You want to **reuse behavior** without inheriting everything. | Avoids tight coupling and unintended inheritance.    |
| You want to **replace parts dynamically.**                    | Easier to swap components (e.g., different engines). |
| You need **more flexibility** for future changes.             | Changes in one class won’t break others.             |

```python
# 1. Opening and Closing Files
# Use open() to open a file and close() to close it.
# Syntax:
# file_object = open("filename", "mode")
# file_object.close()
# open() - opens a file and returns a file object.
# close() - closes the file object to free up resources.
# mode - specifies the mode in which the file is opened (r - read, w - write, a - append, etc.)

# Example:
# Open the file in read mode - "myself.txt"
file = open("myself.txt", "r")  # Open file in read mode
# Perform file operations here
# Lets read the content of the file
content = file.read()
print(content)  # Print the content of the file
file.close()  # Close the file after operations are done
```

```python
# Various versions of read():
# read() - reads the entire file content as a single string.
# read(n) - reads n characters from the file.
# readline() - reads a single line from the file.
# readlines() - reads all lines and returns them as a list of strings.

# 2. Reading from Files
# Example of read()
file = open("myself.txt", "r")
content = file.read()  # Read entire file
print(content)
file.close()
```

```python
# Example of read(n)
file = open("myself.txt", "r")
partial_content = file.read(10)  # Read first 10 characters
print(partial_content)
file.close()
```

```python
# Example of readline()
file = open("myself.txt", "r")
line = file.readline()  # Read first line
print(line)
line2 = file.readline()  # Read second line
print(line2)
file.close()
```

```python
# Example of readlines()
file = open("myself.txt", "r")
lines = file.readlines()  # Read all lines into a list
print(lines)  # Print the list of lines
print("----")
# Iterate through lines
for line in lines:
    print(line)
file.close()
```

```python
# 2. Writing to Files
# Use write() or writelines() to write data to files.
# write() - writes a string to the file.
# writelines() - writes a list of strings to the file.

# Example of write()
file = open("output.txt", "w")  # Open file in write mode
# If the file doesn't exist, it will be created.
# If the file exists, it will be overwritten.
file.write("Hello, World!\n")  # Write a string to the file
file.write("This is a test file.\n")  # Write another string
file.close()  # Close the file
```

```python
# Example of writelines()
file = open("output.txt", "w")  # Open file in write mode
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]
file.writelines(lines)  # Write a list of strings to the file
file.close()  # Close the file
```

```python
# If I dont want the file to be overwritten, I can open it in append mode:
file = open("output.txt", "a")  # Open file in append mode
file.write("Appending a new line.\n")  # Append a new line to the file
file.close()  # Close the file
```

```python
# Read the file to see the content
file = open("output.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
# Example of Adding two numbers and writing result to file

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
res = num1 + num2
print(f"The sum of {num1} and {num2} is {res}")

# Write the result to a file
file = open("result.txt", "a")  # Open file in append mode
file.write(f"The sum of {num1} and {num2} is {res}\n")  # Write result to file
file.close()  # Close the file
```

```python
# 3. Working with File Paths
# You can use absolute or relative paths to specify file locations.
# Absolute Path - full path from root directory
# Relative Path - path relative to the current working directory

import os

# Get the current working directory
cwd = os.getcwd()
print(f"Current Working Directory: {cwd}")

# Join paths
file_path = os.path.join(cwd, "result.txt")
print(f"File Path: {file_path}")

# Open file using absolute path
file = open(file_path, "r")
content = file.read()
print(content)
file.close()
```

```python
# 4. File Modes
# 'r' - Read (default mode)
# 'w' - Write (creates new file or overwrites existing)
# 'a' - Append (adds to end of file)
# 'b' - Binary mode (for non-text files)
# 't' - Text mode (default, for text files)
# 'r+' - Read and Write
```

```python
# File Object Attributes
# file.closed - Returns True if the file is closed, false otherwise
# file.mode - Returns access mode with which the file was opened
# file.name - Returns name of the file

file = open("result.txt", "r")
print(f"File Name: {file.name}")
print(f"File Mode: {file.mode}")
print(f"Is File Closed? {file.closed}")
file.close()
print(f"Is File Closed after closing? {file.closed}")
```

```python
# 5. Using 'with' Statement
# The with statement automatically handles closing the file for you.
with open("output.txt", "r") as file:
    content = file.read()
    print(content)
# No need to call file.close() — it’s done automatically.
```

```python
# 6. Handling File Exceptions (later)

# 7. File Operations (copy, move, delete)
import shutil
# shutil module provides functions for file operations like copy, move, and delete.
# Copy a file
# Syntax: shutil.copy(source, destination)
shutil.copy("output.txt", "output_copy.txt")  # Copy output.txt to output_copy.txt
# Move a file
# Syntax: shutil.move(source, destination)
shutil.move("output_copy.txt", "moved_output.txt")  # Move output_copy.txt to moved_output.txt

import time
time.sleep(5)  # Pause for 5 seconds to see the files created before deletion
# Delete a file
# Syntax: os.remove(file_path)
import os
os.remove("moved_output.txt")  # Delete the file moved_output.txt
print("File operations completed.")
```
