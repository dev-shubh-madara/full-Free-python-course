```python
#write a python program for arithmetic operations and  logical operator
a = 10
b = 3
print("======ARITHMETIC OPERATORS======\n")
print(f"Addition: {a+b}")
print(f"Subtraction: {a-b}")
print(f"Multiplication: {a*b}")
print(f"Division: {a/b}")
print(f"Floor division: {a//b}")
print(f"Remainder: {a%b}")
print(f"Power: {a**b}")
```

```python
print("======COMPARISON OPERATORS======\n")
print(f"a > b: {a>b}")
print(f"a < b: {a < b}")
print(f"a == b: {a==b}")
print(f"a != b: {a!=b}")
print(f"a >= b: {a>=b}")
print(f"a <= b {a <= b}")
```

```python
print("======LOGICAL OPERATORS======\n")
x = 15
print(f"x > 10 and x < 20: {x > 10 and x < 20}")
print(f"x < 10 or x == 15: {x < 10 or x == 15}")
print(f"not(x > 10): {not(x > 10)}")
```

```python
#Write a python program for menu driven calculator

print("=====SIMPLE CALCULATOR======\n")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division\n")

num1 = float(input("Enter First number: "))
num2 = float(input("Enter Second number: "))
choice = int(input("Enter your choice (1-4)"))

print(f"Number 1: {num1}")
print(f"Number 2: {num2}")
print(f"You have selected the choice: {choice}\n")

print("======Result======")

if choice == 1:
    print(f"Addition: {num1 + num2}")
elif choice == 2:
    print(f"Subtraction: {num1 - num2}")
elif choice == 3:
    print(f"Multiplication: {num1 * num2}")
elif choice == 4:
    if num2 != 0:
        print(f"Division: {num1 / num2}")
    else:
        print("Error: Division by zero is not allowed.")
else:
    print("Invalid choice! please select a valid option (1-4).")
```

```python
while(True):
    print("\n=====CALCULATOR======\n")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit\n")
    num1 = float(input("Enter First number: "))
    num2 = float(input("Enter Second number: "))
    choice = int(input("Enter your choice (1-5): "))

    print(f"Number 1: {num1}")
    print(f"Number 2: {num2}")
    print(f"You have selected the choice: {choice}\n")

    if choice == 5:
        print("Exiting the calculator.")
        break
    if choice not in [1, 2, 3, 4, 5]:
        print("Invalid choice! Please select a valid option (1-5).")
        continue

    if choice == 1:
        print(f"Addition: {num1 + num2}")
        continue
    elif choice == 2:
        print(f"Subtraction: {num1 - num2}")
        continue
    elif choice == 3:
        print(f"Multiplication: {num1 * num2}")
        continue
    elif choice == 4:
        if num2 != 0:
            print(f"Division: {num1 / num2}")
            continue
        else:
            print("Error: Division by zero is not allowed.")
            continue
```
